# Testing the deep with a notebook

This is a throwaway notebook used to check how Jupyter content renders inside the *Fons Voyage* dark theme: input cells, stream output, a `pandas` DataFrame, and a `matplotlib` figure. The numbers below are entirely made up — there's no real biology here, just rectangles.

```python
import matplotlib.pyplot as plt
import pandas as pd

plt.rcParams["figure.dpi"] = 110
pd.set_option("display.precision", 3)
```

## A small DataFrame

Per-region neuron counts and the fraction expressing `Cux2` (an upper-layer marker).

```python
df = pd.DataFrame({
    "region": ["VZ", "SVZ", "IZ", "CP"],
    "neuron_count": [12453, 38221, 4192, 67110],
    "ratio_cux2": [0.04, 0.12, 0.31, 0.78],
})
df
```

```python
total = df["neuron_count"].sum()
print(f"Total neurons sampled: {total:,}")
print(f"Mean Cux2+ ratio:      {df['ratio_cux2'].mean():.2f}")
```

## A figure

Bar chart of the (fictional) per-region neuron counts. The figure should render on a light backing so transparent-canvas plots remain legible against the deep-ocean background.

> If this bar chart is hard to read, the theme override for `.jp-OutputArea-output svg` (or the classic `.output svg`) is the place to look.

```python
fig, ax = plt.subplots(figsize=(5.2, 3.2))
ax.bar(df["region"], df["neuron_count"], color="#1f7a7a")
ax.set_ylabel("neurons sampled")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.tight_layout()
plt.show()
```
